```python
import pandas as pd 
df = pd.read_csv('netflix_titles.csv')
```

```python
import sqlalchemy as sal
engine = sal.create_engine('mssql+pyodbc://LAPTOP-135PO6B1/NetflixELT?driver=ODBC+Driver+17+For+SQL+Server')
conn=engine.connect()
```

```python
from sqlalchemy.types import NVARCHAR
df.to_sql(
    'netflix_raw',
    con=conn,
    index=False,
    if_exists='append',
    dtype={
        'title': NVARCHAR(200),
        'director': NVARCHAR(250),
        'cast': NVARCHAR(800),
        'country': NVARCHAR(150),
        'listed_in': NVARCHAR(100),
        'description': NVARCHAR(500)
    }
)

conn.close()
```
